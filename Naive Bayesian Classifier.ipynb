iii) **Apply Cross Validation technique to address overfitting, and test the training score**

```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import StratifiedKFold

nb_classifier = GaussianNB()


kf = StratifiedKFold(n_splits=4, random_state=42, shuffle=True) #JM: Cross-validation with StratifiedKFold, 4 folds

cv_accuracies = [] # JM: Storing accuracy results for each fold

# JM: Loop for cross validation
for fold, (train_index, test_index) in enumerate(kf.split(X, y), 1):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    nb_classifier.fit(X_train, y_train)
    y_pred = nb_classifier.predict(X_test)
    
    
    fold_accuracy = accuracy_score(y_test, y_pred) # JM: Calculate accuracy for the fold
    cv_accuracies.append(fold_accuracy)

    print(f"Fold {fold} - Accuracy: {fold_accuracy * 100:.2f}%")


print(f"\nCross-Validation Accuracies: {cv_accuracies}") # JM: Printing the cross-validation results
print(f"Mean Accuracy: {np.mean(cv_accuracies) * 100:.2f}%")
print(f"Standard Deviation: {np.std(cv_accuracies) * 100:.2f}%")
```

**Conclusion on the use of Cross validation on the test:** Cross validation reduced the observed training accuracy of the model from 92.50% to 88.50%. This means that the model was overfitting the dataset. This led to Overestimation. The 92.50% accuracy obtained may not generalize well with unseen data.

**v) Creating a drug prescription system based on this model and the dataset we have. I will call it SETU Medical Drugs Prescription System**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
from sklearn.naive_bayes import GaussianNB
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline

data = pd.read_csv('drug200.csv')
df = pd.DataFrame(data)

df['Sex'] = df['Sex'].astype('category')
df['BP'] = df['BP'].astype('category')
df['Cholesterol'] = df['Cholesterol'].astype('category')
df['Drug'] = df['Drug'].astype('category')

df['Sex'] = df['Sex'].str.upper()
df['BP'] = df['BP'].str.upper()
df['Cholesterol'] = df['Cholesterol'].str.upper()
df['Drug'] = df['Drug'].str.upper()


ordinal_encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
onehot_encoder = OneHotEncoder(sparse_output=False)
label_encoder = LabelEncoder()

df['Cholesterol'] = df['Cholesterol'].apply(lambda x: x.upper() if isinstance(x, str) else x)
df['Cholesterol'] = df['Cholesterol'].replace({"NORMAL": 0, "HIGH": 1})

df[['BP', 'Cholesterol']] = ordinal_encoder.fit_transform(df[['BP', 'Cholesterol']])

encoded_sex = onehot_encoder.fit_transform(df[['Sex']])
sex_df = pd.DataFrame(encoded_sex, columns=onehot_encoder.get_feature_names_out(['Sex']))

df['Drug'] = label_encoder.fit_transform(df['Drug'])

df = pd.concat([df, sex_df], axis=1)

X = df.drop(columns=['Drug', 'Sex'])
y = df['Drug']

nb_classifier = GaussianNB()

smote = SMOTE(sampling_strategy='auto', random_state=42)
pipeline = Pipeline([('smote', smote), ('nb_classifier', nb_classifier)])

kf = StratifiedKFold(n_splits=4, shuffle=True, random_state=42)

cv_scores = cross_val_score(pipeline, X, y, cv=kf)

print(f"SMOTE Cross-Validation Scores: {cv_scores}")
print(f"Mean Accuracy with SMOTE: {np.mean(cv_scores):.4f}")
print(f"Standard Deviation with SMOTE: {np.std(cv_scores) * 100:.2f}%")

#JM: Now, processing user input
print('*'*50)
print('Welcome to SETU Medical Drugs Prescription System')
print(f'the system prescription accuracy is {np.mean(cv_scores):.4f} therefore your skills are required and this is not the replacement of your skills')
print('Please enter the patient details below to prescribe the right medicine')
print('*'*50)
def get_user_input():
    #JM: Prompting the user for their input (example structure)
    age = int(input("Enter Age: "))
    sex = input("Enter Sex (M/F): ").upper()  # JM: Change input to uppercase
    bp = input("Enter BP (HIGH/LOW/NORMAL): ").upper()  # JM: Change input to uppercase
    cholesterol = input("Enter Cholesterol (HIGH/NORMAL): ").upper()  # JM: Change input to uppercase
    
    # JM: Validate cholesterol input
    if cholesterol not in ['HIGH', 'NORMAL']:
        print("Invalid Cholesterol value entered. Please enter HIGH, or NORMAL.")
        return None
    
    na_to_k = float(input("Enter Na_to_K: "))
    
    return {'Age': [age], 'Sex': [sex], 'BP': [bp], 'Cholesterol': [cholesterol], 'Na_to_K': [na_to_k]}

# JM: Preprocessing and encoding user input
def preprocess_user_input(user_input):
    if user_input is None:
        return None

    # JM: Convert all user input to uppercase
    user_input_df = pd.DataFrame(user_input)
    user_input_df['Sex'] = user_input_df['Sex'].str.upper()
    user_input_df['BP'] = user_input_df['BP'].str.upper()
    user_input_df['Cholesterol'] = user_input_df['Cholesterol'].str.upper()

    # JM: Validate cholesterol input
    if user_input_df['Cholesterol'].iloc[0] not in ['HIGH', 'NORMAL']:
        print("Invalid Cholesterol value entered. Exiting.")
        return None

    # JM: Apply ordinal encoding for 'BP' and 'Cholesterol' due to their nature
    user_input_df[['BP', 'Cholesterol']] = ordinal_encoder.transform(user_input_df[['BP', 'Cholesterol']])

    # JM: One-hot encode 'Sex'
    encoded_sex_input = onehot_encoder.transform(user_input_df[['Sex']])
    sex_df_input = pd.DataFrame(encoded_sex_input, columns=onehot_encoder.get_feature_names_out(['Sex']))

    # JM: Add one-hot encoded 'Sex' columns back to the input data
    user_input_df = pd.concat([user_input_df, sex_df_input], axis=1)

    # JM: Ensure the user input has the same columns as the training data
    user_input_df = user_input_df[X.columns.tolist()]

    return user_input_df

# JM: Make a prediction for the user input
def make_prediction():
    user_input = get_user_input()
    processed_input = preprocess_user_input(user_input)

    if processed_input is None:
        return

    #JM: Train the pipeline on the entire dataset before making a prediction
    pipeline.fit(X, y)

    # JM: Predict using the trained pipeline
    predicted_class = pipeline.predict(processed_input)

    # JM: Convert the numeric prediction back to the original class labels
    predicted_label = label_encoder.inverse_transform(predicted_class)

    # JM: Output the predicted drug class for the user input
    print(f"The predicted drug class for the given input is: {predicted_label[0]}")

# JM: Do you want to continue?
def repeat_process():
    while True:
        make_prediction()
        repeat = input("Do you want to make another prediction? (yes/no): ").strip().lower()
        if repeat != 'yes':
            print("Exiting the prediction process.")
            break

# JM: Start the repeat process
repeat_process()
```
