```python
import gradio as gr
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load GPT-2 model and tokenizer
model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token  # Fixes attention mask issue

# Function for text generation
def generate_text(prompt, max_length, temperature, top_k, top_p, sampling_method, do_sample, repetition_penalty, num_return_sequences):
    if not prompt.strip():
        return ["Please enter a valid prompt"]

    # Tokenize input
    inputs = tokenizer(prompt, return_tensors="pt", padding=True, truncation=True)
    input_ids = inputs.input_ids
    attention_mask = inputs.attention_mask  # ✅ Ensure attention mask is set

    # Ensure valid top_k and top_p values to prevent scalar errors
    if sampling_method == "Top-k Sampling":
        top_p = 1.0  # Disable Top-p when using Top-k
    else:
        top_k = 0  # Disable Top-k when using Top-p

    # Generate text
    outputs = model.generate(
        input_ids,
        attention_mask=attention_mask,  # ✅ Pass attention mask
        max_length=max_length,
        temperature=temperature,
        top_k=int(top_k),  # Ensure integer
        top_p=float(top_p),  # Ensure float
        do_sample=do_sample,  # ✅ User-controlled
        repetition_penalty=repetition_penalty,  # ✅ Prevent repetition
        num_return_sequences=num_return_sequences,  # ✅ Generate multiple responses
        pad_token_id=tokenizer.eos_token_id
    )

    # Decode outputs
    return [tokenizer.decode(output, skip_special_tokens=True) for output in outputs]

# Function to update sliders dynamically
def update_sliders(sampling_method):
    if sampling_method == "Top-k Sampling":
        return gr.update(visible=True), gr.update(visible=False), 50, 1.0  # Default Top-k value, disable Top-p
    else:
        return gr.update(visible=False), gr.update(visible=True), 0, 0.9  # Default Top-p value, disable Top-k

# Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("# Joel S. Mollel: GPT-2 Text Generator")

    prompt = gr.Textbox(label="Enter Prompt: Word or sentence you want to predict the next word(s)")
    max_length = gr.Slider(1, 200, value=50, step=10, label="Max Length")
    temperature = gr.Slider(0.1, 2.0, value=1.0, step=0.1, label="Temperature")

    sampling_method = gr.Radio(["Top-k Sampling", "Top-p Sampling"], label="Select Your Sampling Method", value="Top-k Sampling")

    top_k_slider = gr.Slider(1, 100, value=50, label="Top-k Sampling", visible=True)
    top_p_slider = gr.Slider(0.1, 1.0, value=0.9, step=0.05, label="Top-p Sampling", visible=False)

    do_sample = gr.Checkbox(label="Enable Sampling", value=True)
    repetition_penalty = gr.Slider(1.0, 2.0, value=1.2, step=0.1, label="Repetition Penalty")
    num_return_sequences = gr.Slider(1, 5, value=1, step=1, label="Number of Responses")

    output_text = gr.Textbox(label="Generated Text", lines=5)

    # Update sliders when sampling method changes
    sampling_method.change(update_sliders, inputs=[sampling_method], outputs=[top_k_slider, top_p_slider, top_k_slider, top_p_slider])

    # Generate text button
    btn = gr.Button("Generate")
    btn.click(generate_text, 
              inputs=[prompt, max_length, temperature, top_k_slider, top_p_slider, sampling_method, do_sample, repetition_penalty, num_return_sequences], 
              outputs=output_text)

demo.launch(share=True)
```
